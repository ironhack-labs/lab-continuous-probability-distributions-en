## Exponential Distribution

1. Using `numpy.random.exponential`, create a function that returns a list of numbers exponentially distributed with the mean of 10. 

1. Use the function to generate two number sequences with the size of 10 and 100.

1. Plot the distributions as histograms with the nubmer of bins as 100.

Your output should look like below:

```python
# your code here

# Function to generate exponentially distributed random numbers
def generate_exponential_numbers(mean, size):
    return np.random.exponential(scale=mean, size=size)

# Generate two sequences of exponentially distributed numbers with mean=10
sequence_1 = generate_exponential_numbers(mean=10, size=10)
sequence_2 = generate_exponential_numbers(mean=10, size=100)

# Plotting the histograms with 100 bins
plt.figure(figsize=(12, 6))

# Histogram for the first sequence (size=10)
plt.subplot(1, 2, 1)
plt.hist(sequence_1, bins=100, edgecolor='black')
plt.title('Exponential Distribution (size=10)')
plt.xlabel('Value')
plt.ylabel('Count')

# Histogram for the second sequence (size=100)
plt.subplot(1, 2, 2)
plt.hist(sequence_2, bins=100, edgecolor='black')
plt.title('Exponential Distribution (size=100)')
plt.xlabel('Value')
plt.ylabel('Count')

plt.tight_layout()
plt.show()
```

How are the two distributions different?

The mean changes, so the distribution changes as well.
