## Uniform Distribution

To generate uniform random numbers between any two given values using scipy, we can either use the following code or the code that we have
discussed in class:

```python
# your code here
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform

# Function to generate uniform random numbers
def generate_uniform(bottom, ceiling, count):
    """
    Generate 'count' uniform random numbers between 'bottom' and 'ceiling'
    """
    return uniform.rvs(loc=bottom, scale=ceiling - bottom, size=count)

# Generate two sets of random numbers
numbers1 = generate_uniform(bottom=10, ceiling=15, count=100)
numbers2 = generate_uniform(bottom=10, ceiling=60, count=1000)

# Plot histograms
plt.figure(figsize=(12, 5))

# First histogram
plt.subplot(1, 2, 1)
plt.hist(numbers1, bins=10, color='skyblue', edgecolor='black')
plt.title('Uniform Distribution: 100 numbers (10-15)')
plt.xlabel('Value')
plt.ylabel('Count')

# Second histogram
plt.subplot(1, 2, 2)
plt.hist(numbers2, bins=10, color='salmon', edgecolor='black')
plt.title('Uniform Distribution: 1000 numbers (10-60)')
plt.xlabel('Value')
plt.ylabel('Count')

plt.tight_layout()
plt.show()
```

## Exponential Distribution

1. Using `numpy.random.exponential`, create a function that returns a list of numbers exponentially distributed with the mean of 10. 

1. Use the function to generate two number sequences with the size of 10 and 100.

1. Plot the distributions as histograms with the nubmer of bins as 100.

Your output should look like below:

![exponential distribution](ed.png)

```python
# your code here
import numpy as np
import matplotlib.pyplot as plt

# Function to generate exponentially distributed numbers
def generate_exponential(mean, count):
    """
    Generate 'count' numbers from an exponential distribution
    with specified mean.
    """
    # numpy's scale parameter = mean
    return np.random.exponential(scale=mean, size=count)

# Generate two sequences
exp_seq_10 = generate_exponential(mean=10, count=10)
exp_seq_100 = generate_exponential(mean=10, count=100)

# Plot histograms
plt.figure(figsize=(12,5))

# Histogram for count=10
plt.subplot(1, 2, 1)
plt.hist(exp_seq_10, bins=100, color='skyblue', edgecolor='black')
plt.title('Exponential Distribution: 10 numbers, mean=10')
plt.xlabel('Value')
plt.ylabel('Count')

# Histogram for count=100
plt.subplot(1, 2, 2)
plt.hist(exp_seq_100, bins=100, color='salmon', edgecolor='black')
plt.title('Exponential Distribution: 100 numbers, mean=10')
plt.xlabel('Value')
plt.ylabel('Count')

plt.tight_layout()
plt.show()
```

How are the two distributions different?
The two exponential sequences differ mainly in sample size: the 10-number sequence is very rough and uneven, while the 100-number sequence is smoother and better shows the characteristic exponential decay. The smaller sample has high variability and may not reflect the true distribution, whereas the larger sample approximates the theoretical shape more accurately.

The mean changes, so the distribution changes as well.

What is the probability that the customer will spend more than 15 minutes

```python
# your code here
from scipy.stats import expon

# Parameters
mean_time = 10          # mean = 10 minutes
lam = 1 / mean_time     # lambda = 1/mean
x = 15                  # time of interest

# Compute probability
prob = expon.cdf(x, scale=mean_time)  # scale = 1/lambda = mean
print(f"Probability that a customer spends less than 15 minutes: {prob:.4f}")
```
